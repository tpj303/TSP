```python
# 判決書爬蟲 - 清潔版（修正語法錯誤）
import time
import re
import os

# 先嘗試修復套件問題
def fix_packages():
    """修復套件相依性問題"""
    import subprocess
    import sys

    commands = [
        "pip uninstall -y urllib3 seleniumbase selenium",
        "pip install urllib3==1.26.18",
        "pip install selenium==4.15.2",
        "pip install seleniumbase==4.21.6"
    ]

    for cmd in commands:
        try:
            print(f"執行: {cmd}")
            subprocess.run(cmd, shell=True, check=True)
        except Exception as e:
            print(f"錯誤: {e}")

def test_imports():
    """測試必要的模組導入"""
    try:
        from selenium import webdriver
        from selenium.webdriver.chrome.options import Options
        from selenium.webdriver.common.by import By
        from selenium.webdriver.support.ui import WebDriverWait
        from selenium.webdriver.support import expected_conditions as EC
        from selenium.common.exceptions import TimeoutException, NoSuchElementException
        print("✓ 所有必要模組導入成功")
        return True
    except ImportError as e:
        print(f"✗ 模組導入失敗: {e}")
        return False

class SimpleJudgementCrawler:
    def __init__(self, headless=True):
        """初始化爬蟲"""
        if not test_imports():
            raise ImportError("無法導入必要的模組")

        from selenium import webdriver
        from selenium.webdriver.chrome.options import Options
        from selenium.webdriver.support.ui import WebDriverWait

        print("正在初始化瀏覽器...")

        # Chrome 選項
        chrome_options = Options()
        if headless:
            chrome_options.add_argument('--headless')
        chrome_options.add_argument('--no-sandbox')
        chrome_options.add_argument('--disable-dev-shm-usage')
        chrome_options.add_argument('--disable-gpu')
        chrome_options.add_argument('--window-size=1920,1080')

        # 初始化 WebDriver
        try:
            self.driver = webdriver.Chrome(options=chrome_options)
        except Exception as e:
            print(f"Chrome 初始化失敗: {e}")
            try:
                from webdriver_manager.chrome import ChromeDriverManager
                from selenium.webdriver.chrome.service import Service
                service = Service(ChromeDriverManager().install())
                self.driver = webdriver.Chrome(service=service, options=chrome_options)
            except Exception as e2:
                print(f"使用 webdriver-manager 也失敗: {e2}")
                raise

        self.wait = WebDriverWait(self.driver, 20)
        print("✓ 瀏覽器初始化成功")

    def close(self):
        """關閉瀏覽器"""
        if hasattr(self, 'driver') and self.driver:
            self.driver.quit()
            print("✓ 瀏覽器已關閉")

    def search_judgements(self, keyword, max_results=5):
        """搜尋判決書"""
        from selenium.webdriver.common.by import By
        from selenium.webdriver.support import expected_conditions as EC
        from selenium.common.exceptions import TimeoutException

        try:
            print(f"正在搜尋關鍵字: {keyword}")

            # 載入司法院網站
            self.driver.get('https://judgment.judicial.gov.tw/FJUD/default.aspx')
            time.sleep(3)

            # 輸入搜尋關鍵字
            search_input = self.wait.until(EC.presence_of_element_located((By.ID, "txtKW")))
            search_input.clear()
            search_input.send_keys(keyword)

            # 點擊搜尋
            search_button = self.driver.find_element(By.ID, "btnSimpleQry")
            search_button.click()
            time.sleep(5)

            # 切換到結果框架
            try:
                self.wait.until(EC.frame_to_be_available_and_switch_to_it((By.ID, "iframe-data")))
                time.sleep(2)
            except TimeoutException:
                print("✗ 找不到結果框架")
                return []

            # 獲取判決書連結
            judgement_links = []
            try:
                links = self.driver.find_elements(By.XPATH, "//table[@id='jud']//a")
                for link in links[:max_results]:
                    href = link.get_attribute("href")
                    title = link.get_attribute("title") or link.text.strip()
                    if href and 'FJUD' in href:
                        judgement_links.append({'url': href, 'title': title})

                print(f"✓ 找到 {len(judgement_links)} 個判決書連結")

            except Exception as e:
                print(f"✗ 獲取連結時發生錯誤: {e}")

            # 回到主框架
            self.driver.switch_to.default_content()
            return judgement_links

        except Exception as e:
            print(f"✗ 搜尋時發生錯誤: {e}")
            return []

    def download_judgement(self, judgement_info):
        """下載單一判決書"""
        try:
            url = judgement_info['url']
            title = judgement_info.get('title', 'unknown')

            if '&id=' not in url:
                print(f"✗ 無效的 URL")
                return False

            # 提取 ID
            judgement_id = url.split('&id=')[1].split('&')[0]
            text_url = f'https://judgment.judicial.gov.tw/EXPORTFILE/reformat.aspx?type=JD&id={judgement_id}'

            print(f"正在下載: {title[:50]}...")

            # 載入判決書頁面
            self.driver.get(text_url)
            time.sleep(3)

            # 獲取並處理內容
            content = self.driver.page_source
            text = re.sub(r'<[^>]+>', '\n', content)
            text = text.replace('&nbsp;', ' ')
            text = re.sub(r'\n+', '\n', text).strip()

            # 找到判決書內容
            content_start = None
            for marker in ['裁判字號：', '案號：', '判決']:
                if marker in text:
                    content_start = text.split(marker, 1)[1]
                    break

            if not content_start:
                print(f"✗ 找不到判決書內容")
                return False

            # 準備檔名
            safe_title = re.sub(r'[<>:"/\\|?*]', '_', title)[:50]
            if not safe_title.strip():
                safe_title = f"judgment_{judgement_id}"

            # 儲存檔案
            os.makedirs('judgement_docs', exist_ok=True)
            filename = f'judgement_docs/{safe_title}_{judgement_id}.txt'

            with open(filename, 'w', encoding='utf-8') as f:
                f.write(f"標題: {title}\n")
                f.write(f"ID: {judgement_id}\n")
                f.write(f"來源: {text_url}\n")
                f.write("=" * 50 + "\n")
                f.write(content_start[:10000])  # 限制長度

            print(f"✓ 成功儲存: {os.path.basename(filename)}")
            return True

        except Exception as e:
            print(f"✗ 下載判決書時發生錯誤: {e}")
            return False

def run_test():
    """執行測試"""
    crawler = None
    try:
        print("=== 判決書爬蟲測試 ===")

        # 初始化
        crawler = SimpleJudgementCrawler(headless=True)

        # 搜尋
        keyword = "詐欺"
        results = crawler.search_judgements(keyword, max_results=3)

        if not results:
            print("✗ 沒有找到相關判決書")
            return

        # 下載
        success_count = 0
        for i, judgement in enumerate(results, 1):
            print(f"\n--- 處理第 {i}/{len(results)} 個判決書 ---")
            if crawler.download_judgement(judgement):
                success_count += 1
            time.sleep(2)

        print(f"\n=== 完成 ===")
        print(f"成功下載: {success_count}/{len(results)} 個判決書")
        print("檔案在 'judgement_docs' 資料夾中")

    except Exception as e:
        print(f"✗ 測試失敗: {e}")
    finally:
        if crawler:
            crawler.close()

def custom_search(keyword, max_results=5):
    """自定義搜尋"""
    crawler = None
    try:
        crawler = SimpleJudgementCrawler(headless=True)
        results = crawler.search_judgements(keyword, max_results)

        success_count = 0
        for judgement in results:
            if crawler.download_judgement(judgement):
                success_count += 1
            time.sleep(1)

        print(f"關鍵字: {keyword}, 成功: {success_count}/{len(results)}")

    except Exception as e:
        print(f"搜尋錯誤: {e}")
    finally:
        if crawler:
            crawler.close()

# 主程式
if __name__ == "__main__":
    print("判決書爬蟲程式 v2.0")
    print("如果是第一次使用，可能需要執行 fix_packages()")

    # 執行測試
    run_test()

    # 或使用自定義搜尋：
    # custom_search("竊盜", 5)
```
